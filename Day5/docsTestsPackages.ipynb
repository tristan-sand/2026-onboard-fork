**Problem 1b**

Create docstrings for `Universe`. You need to build a class-level docstring, *as well as* docstrings for each method.

```python
class Universe:
    """
    WRITE THE DOC STRINGS
    """

    def __init__(self):
        self.bodies = []

    def add_body(self, body):
        """
        WRITE THE DOC STRINGS
        """
        self.bodies.append(body)

    def compute_acceleration(self):
        """
        WRITE THE DOC STRINGS
        """
        for body in self.bodies:
            body.a = np.zeros_like(body.r)
        for i, bi in enumerate(self.bodies):
            for j, bj in enumerate(self.bodies):
                if i != j:
                    rij = bj.r - bi.r
                    dist = sum(rij**2) ** 0.5
                    if dist > 0:
                        bi.a += CONFIG['G'] * bj.m * rij / dist**3

    def leapfrog_step(self, dt):
        """
        WRITE THE DOC STRINGS
        """
        for body in self.bodies:
            body.rdot += 0.5 * dt * body.a
        for body in self.bodies:
            body.r += dt * body.rdot
        self.compute_acceleration()
        for body in self.bodies:
            body.rdot += 0.5 * dt * body.a
```
