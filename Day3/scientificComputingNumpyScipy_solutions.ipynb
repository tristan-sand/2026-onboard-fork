### Why Use NumPy for Scientific Computing?

- NumPy is the **foundation of numerical computing** in Python.


- Provides **efficient array operations** and **broadcasting**.


- Enables **vectorized computations** that are faster and more readable than pure Python loops.


It is well known that python is not great at `for` loops. `NumPy` performs basic array operations much faster than a for loop or list comprehension.

```python
import numpy as np
import time

size = 1_000_000
a = list(range(size))
b = list(range(size))

# Pure Python loop
start = time.time()
c = np.empty_like(a)
for num, (a1, b1) in enumerate(zip(a, b)):
    c[num] = a1 + b1
end = time.time()
print(f"for loop time: {end - start:.4f} seconds")

# List comprehension
start = time.time()
c = [a[i] + b[i] for i in range(size)]
end = time.time()
print(f"list comprehension time: {end - start:.4f} seconds")

# NumPy vectorized operation
a_np = np.arange(size)
b_np = np.arange(size)

start = time.time()
c_np = a_np + b_np
end = time.time()
print(f"NumPy time: {end - start:.4f} seconds")
```


### NumPy for Linear Algebra

(*Note* – can leverage duck typing to perform linear algebra on 2D arrays)

- Efficient matrix operations: multiplication, transpose, inverse

- Solving systems of equations

- Eigenvalue decomposition

```python
# Define matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 2]])

# Matrix multiplication
C = A @ B
print(f"Matrix product A @ B:\n{C}")

# Transpose
print(f"Transpose of A:\n{A.T}")

# Determinant
det_A = np.linalg.det(A)
print(f"Determinant of A: {det_A:.2f}")

# Inverse
inv_A = np.linalg.inv(A)
print(f"Inverse of A:\n{inv_A}")

# Eigenvalues and eigenvectors
eigvals, eigvecs = np.linalg.eig(A)
print(f"Eigenvalues: {eigvals}")
print(f"Eigenvectors:\n{eigvecs}")
```

Quick note - when manipulating `numpy` arrays it is very important to know the different between a view and a copy. 


- *view*: Shares memory with the original array. Changes to the view affect the original.
- *copy*: Independent memory. Changes do **not** affect the original array.

- Views are **faster** and more memory-efficient.
- Copies are **safer** when you need to preserve original data.

Here's a weird thing: 

```python
a = np.array([10, 20, 30, 40, 50])

# Create a view
view = a[1:4]
view[0] = 99

print(f"Original array after modifying view: {a}")
print(f"View: {view}")

# Create a copy
copy = a[1:4].copy()
copy[0] = 77

print(f"Original array after modifying copy: {a}")
print(f"Copy: {copy}")
```

### Array Creation

```python
# Creating arrays
a1 = np.array([1, 2, 3])
a2 = np.zeros((2, 3))
a3 = np.ones((3, 3), dtype=float)
a4 = np.arange(0, 10, 2)
a5 = np.linspace(0, 1, 5)
a6 = np.random.default_rng(42).normal(size=(3,4))

print(f'a1: {a1}')
print(f'a2: {a2}')
print(f'a3: {a3}')
print(f'a4: {a4}')
print(f'a5: {a5}')
print(f'a6 shape: {a6.shape}, dtype: {a6.dtype}')
```

### Slicing and Indexing

```python
A = np.arange(1, 10).reshape(3, 3)
print('A =', A)

# Basic slicing
print('First row:', A[0])
print('Last column:', A[:, -1])
print('Center 2x2 block:', A[0:2, 1:3])

# Boolean masking
mask = A % 2 == 0
print('Even mask:', mask)
print('Even elements:', A[mask])

# Note: slices create views (no copy), fancy/boolean indexing create copies
B = A[:, :2]  # view
B[0, 0] = -99
print('After modifying B, A becomes:', A)
```

### Broadcasting and Basic Operations

```python
x = np.array([1., 2., 3.])
y = np.array([[10.], [20.], [30.]])
print('x + y =', x + y)
print('2*x =', 2 * x)
print('x**2 =', x**2)
print('Sum over rows of (x+y):', (x+y).sum(axis=1))

# Broadcasting pitfalls: shape alignment
try:
    _ = np.ones((3,1)) + np.ones((2,))
except ValueError as e:
    print('Broadcasting error example:', e)
```

### Structured Arrays: NumPy's Answer for Wranglin'

Each example so far includes single data types (i.e., homogeneous arrays).  

`numpy` also supports a *compound* dtype, where each element is a full record with named, typed fields.

(Pssst - this was how we did "pandas" before `pandas`)

Structured arrays show up in the following ways:

- When you read certain binary/C-generated files directly into `numpy`.

- Can be faster than a full DataFrame when you don't need all the `pandas` bells and whistles.

- When you see `dtype=[('name', ...), ...]` in legacy code.

```python
dtype = np.dtype([
    ('station', 'U10'),   # unicode string, max 10 characters
    ('temp_c', 'f8'),     # 64-bit float
    ('humidity', 'f8'),   # 64-bit float
    ('n_readings', 'i4')  # 32-bit int
])

data = np.array([
    ('north', 18.2, 55.0, 120),
    ('south', 22.5, 40.0, 118),
    ('east',  19.8, 62.5, 121),
    ('west',  25.1, 33.0, 119),
], dtype=dtype)

print(data)
print(f"\ndtype: {data.dtype}")
print(f"shape: {data.shape}")  # still 1D! four records, not a 4x4 grid
```

In this example, fields are like columns: `data['temp_c']` pulls the whole column as an ordinary 1D array, while `data[0]` pulls a single row as one record.

```python
print(f"All temperatures: {data['temp_c']}")
print(f"First row (a single record): {data[0]}")
print(f"First row's station name: {data[0]['station']}")

# data['temp_c'] is a plain ndarray, so all the usual numpy tools just work
print(f"\nMean temperature: {data['temp_c'].mean():.1f} C")
print(f"Warmest station: {data['station'][np.argmax(data['temp_c'])]}")
```

This looks a bit like `pandas` because `DataFrame`s are just nicer versions of a structured array.

It is easy to go back and forth between the two if needed:

```python
import pandas as pd

# structured array -> DataFrame: column layout already matches, so this is cheap
df = pd.DataFrame(data)
print(df)
print(f"\ndf dtypes:\n{df.dtypes}")

# DataFrame -> structured array (e.g., handing data off to a C/Fortran routine,
# or writing it with a fixed binary layout via np.memmap)
back_to_array = df.to_records(index=False)
print(f"\nBack to a structured array: {back_to_array}")
print(f"Type: {type(back_to_array)}")
```

### Where - my very favorite part of `numpy`

Boolean masks filter arrays, producing an output that is smaller than the input. 

`array[mask]`

What if you want to select *and* replace?

Then we need to transform elements based on some condition while maintaining the same shape. 

- `np.where(condition, value_if_true, value_if_false)` — a vectorized `if / else`.


- `np.select([cond1, cond2, ...], [choice1, choice2, ...], default=...)` — a vectorized `if / elif / elif / else` for more than two outcomes.

```python
# np.where: vectorized if/else, same shape in and out
# flag any station reporting fewer than 120 readings as needing a check
status = np.where(data['n_readings'] < 120, 'check', 'ok')
print(f"Station status: {status}")

# it also works as a plain elementwise conditional, not just for labels
# e.g., floor any humidity reading below 35%
humidity_floored = np.where(data['humidity'] < 35, 35.0, data['humidity'])
print(f"Humidity, floored at 35%: {humidity_floored}")
```

```python
# np.select: vectorized if/elif/elif/else for more than two outcomes
conditions = [
    data['temp_c'] < 15,
    (data['temp_c'] >= 15) & (data['temp_c'] < 22),
    data['temp_c'] >= 22
]
labels = ['cold', 'mild', 'warm']

temp_category = np.select(conditions, labels, default='unknown')
print(f"Temperature category by station: {temp_category}")

# tie it back into the DataFrame from the last section — this is how you'd
# actually use it in practice, as a new derived column
df['temp_category'] = temp_category
print(f"\n{df[['station', 'temp_c', 'temp_category']]}")
```

#### Physical constants

- Provides **precise values** for hundreds of physical constants.
- Includes **unit conversions** for pressure, energy, temperature, etc.


```python
from scipy import constants

# Access physical constants
print(f"Speed of light: {constants.c} m/s")
print(f"Planck constant: {constants.h} J·s")
print(f"Gravitational constant: {constants.G:.3e} m^3·kg^-1·s^-2") # update the Orrery!!

# Convert units
print(f"1 electronvolt in joules: {constants.eV} J")
print(f"1 atmosphere in pascals: {constants.atm} Pa")
```

### Interpolation

`scipy.interpolate` provides multiple useful methods for interpolation: 


- `interp1d`: 1D interpolation (linear, cubic, etc.)
- `griddata`: Interpolation on scattered data
- `UnivariateSpline`: Spline fitting with smoothing


```python
from scipy import interpolate

# Sample data
x = np.linspace(0, 10, 10)
y = np.sin(x)

# Create interpolator
f_interp = interpolate.interp1d(x, y, kind='cubic')

# Evaluate at new points
x_new = np.linspace(0, 10, 100)
y_new = f_interp(x_new)

print(f"Interpolated values at x=6.283: {f_interp(6.283):.4f}")
```

#### Integration

`scipy.integration` enables basic (numeric) calculus:

- `quad`: Integrate a function over a range
- `dblquad`, `tplquad`: Double and triple integrals
- `solve_ivp`: Solve initial value problems for OD


```python
from scipy import integrate

# Define a function to integrate
def f(x):
    return np.sin(x)

result, error = integrate.quad(f, 0, np.pi) # Integrate from 0 to pi
print(f"Integral of sin(x) from 0 to pi: {result:.4f}, error estimate: {error:.2e}")
```
